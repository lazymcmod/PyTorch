```python
import torch

#inputs
x = torch.tensor(6.7) # Input feature
y = torch.tensor(0.0) # True label (binary)

w = torch.tensor(1.0) # weight
b = torch.tensor(0.0) # Bias
```

```python
# Binary Cross-Entropy Loss of scaler
def binary_cross_entoropy_loss(predication,target):
    epsilon = 1e-8 # To prevent log(0)
    predication = torch.clamp(predication,epsilon,1-epsilon)
    return -(target * torch.log(predication) + (1 - target) * torch.log(1-predication))
```

```python
# Forward Pass 
z = w * x + b # Weighted sum (linear parts)
y_pred = torch.sigmoid(z) # Predicted probability

#Compute binary cross_entropy loss
loss = binary_cross_entoropy_loss(y_pred,y)
print(loss)
```

```python
# Derivative:
# 1 dl/d(y_pred): loss with respect to the prediction(y_pred)

dloss_dy_pred = (y_pred - y)/(y_pred*(1-y_pred))

# 2 dy_pred/dz: prediction (y_pred) with respect to z (sigmoid derivative)
dy_pred_dz = y_pred * (1 - y_pred)

# 3 dz/dw and dz/db: z with respect to w and b
dz_dw = x
dz_db = 1

dl_dw = dloss_dy_pred * dy_pred_dz * dz_dw
dl_db = dloss_dy_pred * dy_pred_dz * dz_db

print(f"Manual Gradient of loss w.r.t weight (dw):",{dl_dw})
print(f"Manual Gradient of loss w.r.t bias (db):",{dl_db})
```

```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)

w = torch.tensor(1.0,requires_grad=True)
b = torch.tensor(0.0,requires_grad=True)
z = w * x + b

print(w)
print(b)
print(z)
```
